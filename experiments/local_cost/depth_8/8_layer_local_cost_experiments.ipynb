## All Results Summary

```python
import json
import numpy as np
from pathlib import Path

results_dir = Path('results/local_cost/depth_8')
seeds = [42, 123, 456, 789, 101112]

print('=' * 80)
print('8-LAYER LOCAL COST RESULTS SUMMARY')
print('=' * 80)
print(f'{"Seed":<10} {"Test Acc (%)":<14} {"Train Acc (%)":<14} {"Grad Norm":<15} {"BP Detected":<12} {"Time (s)":<10}')
print('-' * 80)

accs = []
for seed in seeds:
    metrics_file = results_dir / f'seed_{seed}' / 'metrics.json'
    if metrics_file.exists():
        with open(metrics_file) as f:
            m = json.load(f)
        acc = m['test_acc'] * 100
        accs.append(acc)
        print(f'{seed:<10} {acc:<14.2f} {m["final_train_acc"]*100:<14.2f} {m["gradient_stats"]["mean_norm"]:<15.6e} {str(m["barren_plateau_detected"]):<12} {m["training_time"]:<10.1f}')
    else:
        print(f'{seed:<10} NOT FOUND')

if accs:
    print('-' * 80)
    print(f'{"MEAN":<10} {np.mean(accs):<14.2f}')
    print(f'{"STD":<10} {np.std(accs):<14.2f}')
print('=' * 80)

# Verify all files saved
print('\nResults files verification:')
for seed in seeds:
    seed_dir = results_dir / f'seed_{seed}'
    metrics_ok = (seed_dir / 'metrics.json').exists()
    plot_ok = (seed_dir / 'training_history.png').exists()
    status = '\u2705' if (metrics_ok and plot_ok) else '\u274c'
    print(f'  {status} seed_{seed}/  metrics.json={metrics_ok}  training_history.png={plot_ok}')
```
