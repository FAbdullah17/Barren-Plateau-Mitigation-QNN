# local - cost — Layer 6 (5 Individual Runs)

Run each cell independently to execute a single baseline experiment for layer 6.
Results are written to results/baseline/depth_6/seed_{seed}/

```python
import os
import subprocess
import sys
from pathlib import Path

os.chdir('/mnt/d/Hybrid-QNNs')

def run_exp(approach, depth, seed):
    """Run single experiment and stream live stdout/stderr."""
    cmd = ['python', f'experiments/run_{approach}.py',
           f'configs/{approach}_{depth}layer.yaml', '--seed', str(seed)]
    print(f' Running: {approach.upper()} (Layer {depth}, Seed {seed})\n')

    env = os.environ.copy()
    env['PYTHONUNBUFFERED'] = '1'

    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                         text=True, bufsize=1, env=env)
    try:
        for raw_line in p.stdout:
            clean = raw_line.replace('\r', '\n')
            for sub in clean.splitlines(keepends=True):
                print(sub, end='')
                sys.stdout.flush()
        p.wait()
    except KeyboardInterrupt:
        p.kill()
        raise

    if p.returncode == 0:
        print(f'\n SUCCESS — results saved to results/{approach}/depth_{depth}/seed_{seed}/')
    else:
        print(f'\n FAILED (exit code {p.returncode})')
```

## Run — Seed 42

```python
run_exp('local_cost', 6, 42)
```
