# Individual Experiment Runner

Run experiments individually or sequentially
- **3 Layers**: 4, 6, 8
- **3 Approaches per layer**: Baseline, Layerwise, Local_cost
- **5 Seeds per approach**: 42, 123, 456, 789, 101112
- **Total**: 45 cells

```python
import os
import subprocess
from pathlib import Path

os.chdir('/mnt/d/Hybrid-QNNs')

def run_exp(approach, depth, seed):
    """Run single experiment and display live progress"""
    cmd = ['python', f'experiments/run_{approach}.py', 
           f'configs/{approach}_{depth}layer.yaml', '--seed', str(seed)]
    print(f'Running: {approach.upper()} (Layer {depth}, Seed {seed})\n')
    result = subprocess.run(cmd, text=True)
    print(f'\n{"SUCCESS" if result.returncode == 0 else "❌ FAILED"} - Results saved to results/{approach}/depth_{depth}/seed_{seed}/')
```
